```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

experiment = pd.DataFrame({
    "group": ["A"] * 1000 + ["B"] * 1000,
    "started_movie": np.concatenate([
        rng.binomial(1, 0.08, size=1000),
        rng.binomial(1, 0.10, size=1000),
    ]),
})

summary = experiment.groupby("group")["started_movie"].agg(
    goren_kisi="size",
    baslatan_kisi="sum",
)
summary["start_rate_pct"] = (
    100 * summary["starts"] / summary["viewers"]
).round(2)

summary
```

```python
n_a = summary.loc["A", "viewers"]
n_b = summary.loc["B", "viewers"]

rate_a = summary.loc["A", "starts"] / n_a
rate_b = summary.loc["B", "starts"] / n_b
difference = rate_b - rate_a

standard_error = np.sqrt(
    rate_a * (1 - rate_a) / n_a + rate_b * (1 - rate_b) / n_b
)

lower_bound = 100 * (difference - 1.96 * standard_error)
upper_bound = 100 * (difference + 1.96 * standard_error)

print(f"Gözlenen fark: {100 * difference:.1f} yüzde puan")
print(f"%95 güven aralığı: {lower_bound:.2f} ile {upper_bound:.2f} yüzde puan")
```

In the simulated A/B test, the movie-start rate was 7.8% for cover A and 11.1% for cover B. Cover B's observed advantage was 3.3 percentage points. The 95% confidence interval ranged from 0.74 to 5.86 percentage points, favoring B. However, if switching covers required an improvement of at least 2 percentage points, this experiment alone would not be sufficient to make that decision.
